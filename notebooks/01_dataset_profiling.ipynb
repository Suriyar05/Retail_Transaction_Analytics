```python
import pandas as pd

file_path = "../data/raw/Online_Retail_II.xlsx"

df_2009_2010 = pd.read_excel(
    file_path,
    sheet_name="Year 2009-2010"
)

df_2010_2011 = pd.read_excel(
    file_path,
    sheet_name="Year 2010-2011"
)

print("Year 2009-2010")
print("Shape:", df_2009_2010.shape)
print("Columns:", df_2009_2010.columns.tolist())

print("\nYear 2010-2011")
print("Shape:", df_2010_2011.shape)
print("Columns:", df_2010_2011.columns.tolist())
```

```python
print("=== Year 2009-2010 ===")
display(df_2009_2010.head())

print("=== Year 2010-2011 ===")
display(df_2010_2011.head())
```

```python
print("=== 2009-2010 Data Types ===")
print(df_2009_2010.dtypes)

print("\n=== 2010-2011 Data Types ===")
print(df_2010_2011.dtypes)
```

```python
print("=== 2009-2010 ===")
print("Shape:", df_2009_2010.shape)

print("\n=== 2010-2011 ===")
print("Shape:", df_2010_2011.shape)

print("\n=== 2009-2010 Summary ===")
display(df_2009_2010.describe(include="all"))

print("\n=== 2010-2011 Summary ===")
display(df_2010_2011.describe(include="all"))
```

```python
print("=== Missing Values: 2009-2010 ===")
print(df_2009_2010.isnull().sum())

print("\n=== Missing Values: 2010-2011 ===")
print(df_2010_2011.isnull().sum())
```

```python
print("=== Duplicate Rows ===")
print("2009-2010:", df_2009_2010.duplicated().sum())
print("2010-2011:", df_2010_2011.duplicated().sum())
```

```python
print("=== Date Range ===")

print(
    "2009-2010:",
    df_2009_2010["InvoiceDate"].min(),
    "to",
    df_2009_2010["InvoiceDate"].max()
)

print(
    "2010-2011:",
    df_2010_2011["InvoiceDate"].min(),
    "to",
    df_2010_2011["InvoiceDate"].max()
)
```

```python
print("=== Negative Quantity ===")

print("2009-2010:", (df_2009_2010["Quantity"] < 0).sum())
print("2010-2011:", (df_2010_2011["Quantity"] < 0).sum())

print("\n=== Zero Quantity ===")

print("2009-2010:", (df_2009_2010["Quantity"] == 0).sum())
print("2010-2011:", (df_2010_2011["Quantity"] == 0).sum())

print("\n=== Negative Price ===")

print("2009-2010:", (df_2009_2010["Price"] < 0).sum())
print("2010-2011:", (df_2010_2011["Price"] < 0).sum())

print("\n=== Zero Price ===")

print("2009-2010:", (df_2009_2010["Price"] == 0).sum())
print("2010-2011:", (df_2010_2011["Price"] == 0).sum())
```

```python
negative_qty_2009 = df_2009_2010[
    df_2009_2010["Quantity"] < 0
]

negative_qty_2010 = df_2010_2011[
    df_2010_2011["Quantity"] < 0
]

print("Sample negative transactions - 2009-2010")
display(negative_qty_2009.head(10))

print("Sample negative transactions - 2010-2011")
display(negative_qty_2010.head(10))
```

```python
print(
    negative_qty_2009["Invoice"].astype(str).str[0].value_counts()
)

print(
    negative_qty_2010["Invoice"].astype(str).str[0].value_counts()
)
```
